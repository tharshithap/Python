## IF CONDITIONS

```python
# 26. If age >= 18 print "Eligible to Vote"
d = {'name':'Sai', 'age':22}
if d.get('age', 0) >= 18:
    print("Eligible to Vote")
```

```python
# 27. Pass or Fail based on marks
d = {'name':'Rahul', 'marks':65}
if d.get('marks', 0) >= 40:
    print("Pass")
else:
    print("Fail")
```

```python
# 28. High Salary check
d = {'salary':70000}
if d.get('salary', 0) > 50000:
    print("High Salary")
```

```python
# 29. If course is "Python", print "ML Eligible"
d = {'course':'Python'}
if d.get('course') == 'Python':
    print("ML Eligible")
```

```python
# 30.  Print email if key exists, else "Email Not Available"
d = {'name':'Sai'}
print(d.get('email', 'Email Not Available'))
```

## NESTED IF

```python
# 31. Check pass / distinction
d = {'marks':82}
marks = d.get('marks', 0)
if marks >= 40:
    if marks >= 75:
        print("Distinction")
    else:
        print("Pass")
```

```python
# 32. Age and Aadhaar check
d = {'age':20, 'aadhaar': True}
if d.get('age', 0) >= 18:
    if d.get('aadhaar'):
        print("Eligible for Driving License")
    else:
        print("Aadhaar Required")
```

```python
# 33. Username and password check
d = {'username':'admin', 'password':'python123'}
if 'username' in d and d.get('password') == 'python123':
    print("Login Successful")
```

```python
# 34. Experience and rating for promotion eligibility
d = {'experience':3, 'rating':5}
if d.get('experience', 0) >= 2:
    if d.get('rating', 0) >= 4:
        print("Promotion Eligible")
    else:
        print("Needs Improvement")
```

```python
# 35. Scholarship system (marks + income)
d = {'marks':82, 'income':250000}
if d.get('marks', 0) >= 60:
    if d.get('income', float('inf')) <= 300000:
        print("Scholarship Approved")
    else:
        print("Income Too High")
else:
    print("Marks Not Eligible")
```
