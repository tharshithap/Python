# 5.Find the sum of numbers from 1 to N 


```python
n = int(input("Enter a number: "))
total = 0
for i in range(1, n + 1):
    total = total + i
print("Sum =", total)
```

# 10.Reverse a string


```python
s  = input()
reverse = ""
for ch in s:
    reverse = ch + reverse
print("Reversed String=",reverse )
```

# 11.Count vowels 


```python
s  = input()
vowels = ""
for ch in s:
    if ch.lower() in "aeiou":
        vowels = ch + vowels
print("vowels in String=",vowels )
```

# 12.Count consonants 

```python
s  = input()
constant = ""
for ch in s:
    if ch.lower() not in "aeiou":
        constant = ch + constant
print("constants =", constant)
```

# 13.Find the largest element in a list 

```python
lst = [ 4,5,6,7,8,9,2,56,78,99]
largest = lst[0]
for i in lst:
    if i>largest:
        largest = i
print("Largest number:",largest)
```

# 14.Find the smallest element 


```python
lst = [ 4,5,6,7,8,9,2,56,78,99]
smallest = lst[0]
for i in lst:
    if i<smallest:
        smallest = i
print("Smallest number:",smallest)
```

# 15. Count positive numbers 


```python
n = int(input("How many numbers do you want to enter? "))
count = 0
for i in range(n):
    number = int(input(f"Enter number {i + 1}: "))
    if number > 0:
        print(number, "is positive")
        count += 1
    else:
        print(number, "is not positive")
print("Total positive numbers =", count)
        
```

# 16.Find the sum of even numbers in a list 


```python
lst = [ 12,3,4,57,66,89,43,45]
sum = 0
for i in lst:
    if i % 2==0:
        print(i)
        sum += i
print("Sum of even numbers:",sum)
```

# 17.Check Prime Number


```python
n = int(input("Enter a number: "))

for i in range(2, n):
    if n % i == 0:
        print("Not a prime number")
        break
else:
    print("Prime number")
```
