```python
def calculator():
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    operator = input("Enter operator (+, -, *, /): ")

    if operator == "+":
        return f"Sum of {num1} + {num2} = {num1 + num2}"
    elif operator == "-":
        return f"{num1} - {num2} = {num1 - num2}"
    elif operator == "*":
        return f"{num1} * {num2} = {num1 * num2}"
    elif operator == "/":
            return f"{num1} / {num2} = {num1 / num2}"
    else:
        return "Invalid operator."
```

```python
calculator()
```

# map()
--> If you have a list of elemnts and you want to perform same operation on each & every eloment then we can go with map()

Example:- [1,2,3,4,5]--> add 10

o/p:- [11,12,13,14,15]

len(input) == len(output)## both lengths must bee same

## map syntax()
-> list(map(function,iterables))

```python
lst = [1,2,3,45,67]
```

```python
def add_10(x):
    for i in x:
        result.append(i+10)
    return result
    
```

```python
add_10(lst)
```

```python
def add_10(x):
    return x+10
```

```python
list(map(add_10,lst))
```

```python
fun = lambda x:x+10
```

```python
list(map(fun,lst))
```

```python
list(map(lambda x:x+10,lst))
```

```python
name = ["Tharshitha","Thanmai","Josh kamal","Pavithra"]
# find the len of each name
# convert ecach nam into upper case lettering
#convert each name frist letter capital 
# reverse each name 
# find the names starts with letter t
```

```python
list(map(lambda x: x[::-1],name))
```

```python
list(map(lambda x: x.capitalize(),name))
```

```python
list(map(lambda x: x.upper(),name))
```

```python
list(map(lambda x: len(x),name))
# or 
list(map(len,name))
```

```python
list(map(lambda x:x[0]=="T",name))
```

```python
gender = ["male","female","male","female","male","female"]
list(map(lambda x: 1 if x=="male" else 0,gender))
```

```python
prices = ["1,999/-","2,000/-","1,499/-","1555/-"]
list(map(lambda x: prices(int), prices))
```

```python
prices = ["1,999/-", "2,000/-", "1,499/-", "1555/-"]

result = list(map(lambda x: int(x.replace(",", "").replace("/-", "")), prices))

print(result)
```
