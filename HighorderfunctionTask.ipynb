# 1. E-commerce — Calculate Final Prices

An e-commerce company has product prices:

prices = [1200, 2500, 999, 4500, 800]

A festival discount of 10% is applied to every product.

**Using lambda and map():**

**Expected:**

[1080, 2250, 899.1, 4050, 720]

**Interview follow-up:**
 How would you round all prices to the nearest integer?


```python
prices = [1200, 2500, 999, 4500, 800]
l = list(map(lambda x:x*0.90 , prices))
print(l)
print(round(l))
```

**A food delivery application has delivery distances:**

distances = [2, 5, 8, 10, 15]

**Charging rules:**

First 5 km  → ₹30
Above 5 km  → ₹30 + ₹10 for every additional km

Using map() and lambda, calculate the delivery charge for each order.
    
**Expected:**
    
[30, 30, 60, 80, 130]

# 5. Product Inventory — Find Total Stock Value

You have:

products = [
   (100, 5),
   (250, 10),
   (50, 20),
   (500, 3)
]

**Each tuple represents:**

(price, quantity)

Using map() and lambda, calculate the value of each product's inventory.
Then use reduce() to find the total inventory value.

**Expected:**

[500, 2500, 1000, 1500]

**Total:**

5500

🔥 This is an excellent interview question because it combines map() + reduce() + lambda.


```python
products = [(100, 5),(250, 10),(50, 20),(500, 3)]
total = 0
for i in products:
    values = i[0]*i[1]
    total += values
print(total)
```

```python
from functools import reduce
products = [(100, 5),(250, 10),(50, 20),(500, 3)]
l = list(map(lambda x:x[0]*x[1],products))
print(l)
total = reduce(lambda x,y:x+y,l)
print(total)
```

# 8. Salary Processing — Find Highest Salary

 Why can reduce() solve this but map() cannot directly?

# 9. Transaction System — Separate Valid Transactions
**A payment system receives:**

transactions = [500, 1200, -300, 4500, -100, 7000]

Negative transactions represent invalid entries.
    
Using filter() would be the natural solution, but the interviewer asks:
    
Can you use a combination of map(), lambda, and reduce() to process these transactions?
Then calculate the total of valid transactions.
    
**Expected:**
    
13200
    
This tests whether you understand when each higher-order function is appropriate, not just syntax.


```python
transactions = [500, 1200, -300, 4500, -100, 7000]
pos = list(map(lambda x: x if x>=0 else 0 ,transactions))
print(pos)
total = reduce(lambda x,y: x+y , pos)
print(total)
```

# 12. — Student Result Processing

You have:
    
students = [
   ("Ravi", [80, 75, 90]),
   ("Anita", [90, 85, 95]),
   ("Raj", [60, 70, 65])
]

Using map(), reduce() and lambda:
    
Calculate total marks for each student.
    
Calculate average marks.
    
Find the student with the highest total.
    
This question tests nested higher-order functions.


```python
students = [ ("Ravi", [80, 75, 90]), ("Anita", [90, 85, 95]), ("Raj", [60, 70, 65]) ]
total = list(map(lambda x: reduce(lambda a,b : a+b , x[1]),students))
print(total)
avg = list(map(lambda x: round(x / len(students)), total))
print(avg)
max = reduce(lambda x,y: x if x>y else y,total)
print(max)
```

# 13.  — Bank Balance

Transactions:

transactions = [
   ("deposit", 5000),
   ("withdraw", 1000),
   ("deposit", 2000),
   ("withdraw", 500),
   ("deposit", 3000)
]

**Starting balance:**

₹10,000

Use reduce() with lambda to calculate the final balance.
    
**Expected:**

18500


```python
from functools import reduce

transactions = [
    ("deposit", 5000),
    ("withdraw", 1000),
    ("deposit", 2000),
    ("withdraw", 500),
    ("deposit", 3000)
]

balance = reduce(lambda x, y: x + y[1] if y[0] == "deposit" else x - y[1],transactions,10000)

print(balance)
```

# 15. Expert Challenge — Sales Pipeline

A company has:
    
sales = [12000, 15000, 8000, 25000, 30000, 5000]

Business rules:
    
Ignore sales below ₹10,000.
    
Apply 10% commission to remaining sales.
    
Find total commission.
    
Use only filter(), map(), reduce() and lambda.
    
Expected:
    
8200
    
This is one of the best questions to test whether you understand the complete pipeline:

filter → map → reduce

Most important pattern to remember

filter() → select data

map()    → transform data

reduce() → combine data into one result

For example:

reduce(
   lambda x, y: x + y,
   map(
       lambda x: x * 0.10,
       filter(lambda x: x >= 10000, sales)
   )
)

That filter → map → reduce pattern is especially important in real-world data-processing interview questions.



```python
sales = [12000, 15000, 8000, 25000, 30000, 5000]
sal = list(map(lambda x : 0 if x<=10000 else x*(0.10),sales))
print(sal)
total = reduce(lambda x,y:x+y , sal)
print(total)
```
