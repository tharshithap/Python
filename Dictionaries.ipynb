```python
emp = {
    "emp_id": int(input("Enter the employee id: ")),
    "emp_name": input("Enter the employee name: "),
    "dept": input("Enter the department name: "),
    "age": int(input("Enter the age: ")),
    "Salary": int(input("Enter the salary: ")),
    "Email_id": input("Enter the Email id: ")
}

search_id = int(input("Enter the id to search: "))

if search_id == emp["emp_id"]:
    print("ID:", emp["emp_id"])
    print("Name:", emp["emp_name"])
    print("Dept:", emp["dept"])
    print("Age:", emp["age"])
    print("Salary:", emp["Salary"])
    print("Email:", emp["Email_id"])
else:
    print("Data is not present")
```

```python

name = input("Enter employee name: ")

if name in emp["emp_name"]:
    index = emp["emp_name"].index(name)

    print("Employee ID:", emp["emp_id"][index])
    print("Name:", emp["emp_name"][index])
    print("Age:", emp["age"][index])
    print("Salary:", emp["salary"][index])
else:
    print("Employee not found")
```

# Copy()

```python
import copy

l = l=[[1,2,3,4,5,],[6,7,8,9]]

shallow = copy.copy(l) 

shallow[0][0] = 21

print("Shallow Copy:", shallow)
print("Original List:", l)
```

```python
shallow[0][0] = 21

print(shallow)
print(l)
```

```python
l=[[1,2,3,4,5,],[6,7,8,9]]
deep = copy.deepcopy(l)
print(l)
print(deep)
```

```python
deep[0][0]=21
print(deep)
```
