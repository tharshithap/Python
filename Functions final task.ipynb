```python
#1. Write a function that takes a list and returns the difference between the largest and smallest elements.
def diff_large_smallest(lst):
    return max(lst) - min(lst)
```

```python
print(diff_large_smallest([10,45,23,67,12]))
```

```python
#2. Write a function that takes a list and returns the second smallest element.
def second_smallest(lst):
    a = min(lst)
    lst2 = []
    for x in lst:
        if x > a:
            lst2.append(x)
    return min(lst2)
```

```python
print(second_smallest([4, 1, 7, 2, 1]))
```

```python
#3. Write a function that takes a list and returns elements that appear more than once.
def duplicates(lst):
    org = []
    dup = []
    for x in lst:
        if x in org:
            dup.append(x)
        else:
            org.append(x)
    return list(dup)
```

```python
print(duplicates([1,2,3,4,5,6,2,1]))
```

```python
#4. Write a function that takes a list and returns a list of cumulative sums.
def cumulative_sums(lst):
    total = 0
    result=[]
    for x in lst:
        total +=x
        result.append(total)
    return result
```

```python
print(cumulative_sums([3,4,5,6,7]))
```

```python
#5.write a function that rotates a list to the right by one position.
def rotate_right(lst):
    return [lst[-1]]+lst[:-1]
```

```python
print(rotate_right([2,3,4,5]))
```

```python
#6. Write a function that takes a tuple and returns the product of all elements except the largest.
def pro_elements(t):
    m = max(t)
    prod = 1
    for x in t:
        if x != m:
            prod *= x
    return prod
```

```python
pro_elements((2,3,4,5,6,7))
```

```python
#9. Write a function that takes a tuple and returns the count of elements greater than the average.
def tup_elements(t):
    count=0
    avg = sum(t) / len(t)
    for i in t:
        if i>avg:
            count  += i
    return count
    
    
```

```python
#10. Write a function that takes a dictionary and returns keys whose values are greater than 50.
def greater_elements(d):
    res=[]
    for k in d:
        if d[k] > 50:
            res.append(k)
    return res
```

```python
greater_elements({"a":30, "b":60, "c":80})
```

```python
#11. Write a function that swaps keys and values in a dictionary.
def greater_elements(d):
    res={}
    for k in d:
        res[d[k]]= k
    return res
```

```python
greater_elements({"a":1, "b":2})
```

```python
#12. Write a function that takes a dictionary and returns the key with the maximum value.4
def max_elements(d):
    res={}
    for k in d:
        if d[k] < 4:
            res.append(k)
    return res
```
