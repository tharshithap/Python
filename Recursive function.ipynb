# Recursive Functions
- A function calling itself again and again is called recursion
  
**Benifits**
- No of lines of code will be reduced
  
 **Problems**
- Dubbugging is bit difficult
- Problem of entering into infinte loop
- Bit complex to write and take more time than normal function in some times

```python
count = 0

def fun(x):
    global count
    print(x)
    count += 1
    if x <= 0:
        return count
    return fun(x - 1)
```

```python
fun(10)
```

```python
def reduce(n):
    if n == 0:
        return 0 
    print(n)
    n = n -1
    return reduce(n)
```

```python
reduce(10)
```

# Factiorial of a number


```python
import time
start = time.time()
def fact(num):
    f = 1
    for i in range(1,num+1):
        f = f * i
    return f
fact(100)
end = time.time()
print("total time taken:", end-start)
```

```python
import time
start = time.time()
def fact(n):
    if n == 0 :
        return 1 
    return n*fact(n-1)
fact(5)
end = time.time()
print("total time taken:", end-start)    
```

```python
def fibonacci(n):
    if n==0:
        return n 
    if n == 1:
        return n
    return fibonacci(n-1)+ fibonacci(n-2)
    
```

```python
fibonacci(6)
```

# Add memory

```python
def fib(n,memory={}):
    if n ==0:
        return 0 
    if n == 1:
        return 1
    if n in memory:
        return memory[n]
    else:
        memory[n] = fib(n-1) + fib(n-2)
    return memory[n]
    
```

```python
fib(40)
```
