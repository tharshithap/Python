# Tuples
- declared by '()'
- collection which can be indexed but immutable(i.e-apple,250)
- we cannot modify the elements

# Tuple declaration

```python
t = (10,3.14,True,"Innomatics",65,10,3.14)
print(t)
print(type(t))
```

# Accessing of elements in tuple

```python
print(t[0])
print(t[1])
```

# How to add elements in tuple?

```python
t4= 15,16,17
print(t2+t4)
```
