## Types of Variables
1. **Instance Variable**
    - Instance variable will create inside a class and inside a method
    - we can access and update instance variable using only object
2. **Class/Staticvariable**
   - class variable should create inside the class and outside the methos
   - class variable can access in two ways
     
     A. Using object
     
     B. Using class name

```python
class Types_of_var:
    var1 = 20 ## Class variable
    def __init__(self):
        self.ins_var ="some data"
    def method1(self):
        self.var2=30 # Instance variable
    def method2(self):
        self.var3=40 # Instance variable
        self.var4=50 # Instance variable
        
```

2. **Class method**

   - This 'classmethod' decorator will convert instance method to class method
   - In class method we will use 'cls' as default reference pointer instead of 'self'
   - the main purpose of using class methods are to update the class varialble
