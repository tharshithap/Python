Data Type conversion


```python
#int-->float
x=10
print(x)
print(type(x))
y=float(x)
print(y)
print(type(y))
      
```

```python
#int--->complex
a=90
print(a)
print(type(a))

b=complex(a)
print(b)
print(type(b))
```

```python
#float--->int
f=90.90
print(f)
print(type(f))

i=int(f)
print(i)
print(type(i))
```

```python
#float---->complex
f=90.90
print(f)
print(type(f))

c=complex(f)
print(c)
print(type(c))
```
