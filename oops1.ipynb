```python
class BankData:
    name = "State Bank"
    def __init__(self, name, phone, address,panno,accounttype):        
        self.name = name # method / instance variable
        self.phone = phone
        self.address = address
        self.panno = panno
        self.accounttype= accounttype
        self.balance = 0
    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)
    def withdraw(self, amount):
            if amount<self.balance:
                self.balance -= amount
            else:
                print("Insufficent Balance")
    def summary(self):
        print(f'The account summary : {self.name}--{self.balance}')
        
```

```python
cust1 = BankData("Ravi", "abci3456","9876543210", "Hyderabad","savings")
```

```python
cust1.deposit(5000)
cust1.summary()
cust1.withdraw(4000)
cust1.summary()
```

```python
class school:
    school_name = "Innomatics"
    def __init__(self,noofstud,nooffacu,classes):
        self.student = noofstud
        self.faculty = nooffacu
        self.classes = classes
    def show(self):
        print(f'school summary: class strength-{self.student},faculty strength-{self.faculty},class-{self.classes}')
    @classmethod
    def update_name(cls,new_name):
        cls.school_name = new_name

school1 = school(65,2,10)

school1.show()
```
