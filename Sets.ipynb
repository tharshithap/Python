# Sets
- declared using '{}'
- set is unorderd and unindexed
- set will not allow duplicates
- In set we can't use index and slicing because set is unodered
- We can't add multiple values using add()
- We can use update() to add multiple values

```python
# set declaration
s = {10,20,30,40}
print(s)
print(type(s))
```

```python
# creating a empty set
s1 = set()
print(s1)
print(type(s1))
```

```python
# Create a set using a list
lst = [2,3,4,5,'tharshi',8.5]
print(lst)
print(type(lst))
s2 = set(lst)
print(s2)
print(type(s2))
```

```python
# single value adding use add()
s2.add("thannu")
s2
```

```python
# multiple value adding use update()
s2.update([1,2,3,4,5,6])
print(s2)
```

# Sort()

```python
s1 = {11,12,13,14,15,7}
print(sorted(s1,reverse = True))
print(sorted(s1,reverse = False))
```

# Discard()
- It will discard the value from the set
- If the value is not there in the set, discard will not throw any error

```python
s1.discard(13)
s1
```

```python
s1.discard(16) # even the value is not present it will not throw any error
```

```python
# the remove() function will throw the error
s1.remove(14)
print(s1)
```

```python
s1.remove(16)
print(s1)
```

# pop()
- we can't predict which value is poped

```python
s1.pop()
s1
```

# clear()

```python
s1.clear()
s1
```

# Other set method functions
- union()
- intersection()
- difference()

# Union()

```python
s1 = {1,2,3,4,5,6}
s2 = {4,5,6,7,8,9}
s1.union(s2)
```

# Intersection()

```python
s1 = {1,2,3,4,5,6}
s2 = {4,5,6,7,8,9}
s1.intersection(s2)
```

# difference()

```python
s1 = {1,2,3,4,5,6}
s2 = {4,5,6,7,8,9}
s1.difference(s2)
```

# Symmetric Difference
- It removes the common values

```python
s1 = {1,2,3,4,5,6}
s2 = {4,5,6,7,8,9}
s1.symmetric_difference(s2)
```
