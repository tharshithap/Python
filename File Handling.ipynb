## Expection Handiling


Error:- some issues with the program 
--> Exception is nothing but an error

Types of Errors:-

1.Runtime Error

2.Syntax Error

3.Logical Error

--**Only Runtime errors are come under exception**

**How tao handle Exceptions(Run time errors)**

1.try

2.except

```python
num1 = int(input("Enter a number:"))
num2 = int(input("Enter a number:"))

print(num1/num2)
```

```python
num1 = int(input("Enter a number:"))
num2 = int(input("Enter a number:"))
if num2 !=0:
    print(num1/num2)
else:
    print("Dont pass Zero")
```

```python
## here the try and except will handle the errors and execute the next code of line
## the problemetic code is placed in the try() block
num1 = int(input("Enter a number:"))
num2 = int(input("Enter a number:"))
try:
    print(num1/num2)
except:
    print("you should not pass zero in the denominator")
```

## Finally
- Finally is a keyword will excecute everytime irrespective of exceptions

```python
try:
    with open("./files/doreamon.jpg",mode="r")as file:
        print(file.read())
except FileNotFoundError:
    print("file is not available!.check the file location")
finally:
    print("Something is fishy!")
    
```

```python
contacts = {}

while True:
    print("------ Phone Book ------")
    print("1. Get Contacts")
    print("2. Add Contact")
    print("3. Update Contact")
    print("4. Delete Contact")
    print("5. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        if contacts == 0:
            print("No contacts available!")
        else:
            print("------ Contacts ------")
            for name, phone in contacts.items():
                print("Name:", name, "| Phone:", phone)
    elif choice == "2":
        name = input("Enter contact name: ")
        phone = input("Enter phone number: ")

        contacts[name] = phone
        print("Contact added successfully!")

    elif choice == "3":
        name = input("Enter contact name to update: ")

        if name in contacts:
            phone = input("Enter new phone number: ")
            contacts[name] = phone
            print("Contact updated successfully!")
        else:
            print("Contact not found!")

    elif choice == "4":
        name = input("Enter contact name to delete: ")

        if name in contacts:
            del contacts[name]
            print("Contact deleted successfully!")
        else:
            print("Contact not found!")

    elif choice == "5":
        print("Thank you!")
        break

    else:
        print("Invalid choice!")
```
