# Inheriting one class properties to another class i.e parent class to child class

```python
class a:
    def feature1(self):
        print("this is feature1")


class b(a):   # child class
    def __init__(self, name):
        self.name = name
        print(f"this is {self.name}")

    def feature2(self):
        print("this is feature2")


cust2 = b("penubothu")

cust2.feature2()
cust2.feature1()
```

# Types of Inheritance
- single Inheritance
- multiple Inheritance
- multi-level Inheritance
- Hierarhical Inheritance
- Hybrid Inheritance

# Single Inheritance

```python
class a:
    def __init__(self,name):
        self.name = name
    def feature1(self):
        print("this is feature1")


class b(a):   # child class
    def __init__(self, name):
        self.name = name
        print(f"this is {self.name}")

    def feature2(self):
        print("this is feature2")


cust2 = b("penubothu")

cust2.feature2()
cust2.feature1()
```

# Multiple Inheritance

```python
class a:# parent class
    def __init__(self,name):
        self.name = name
        print(self.name)
    def feature1(self):
        print("this is feature1")


class b:   # parent class
    def __init__(self, name):
        self.name = name
        print(self.name)
    def feature2(self):
        print("this is feature2")
class c(a,b): # child
    def __init__(self,name):
        self.name = name
        print(self.name)
    def feature3(self):
        print("This is feature 3")
cust3 = c("Data Science")
print(cust1.name)
print(cust2.name)
print(cust3.name)
cust3.feature2()
cust3.feature1()
cust3.feature3()
```

# Multilevel Inheritance

```python
class a: # Grand parent class
    def __init__(self,name):
        self.name = name
        print(self.name)
    def feature1(self):
        print("this is feature1")


class b(a):   # parent class
    def __init__(self, name):
        self.name = name
        print(self.name)
    def feature2(self):
        print("this is feature2")
class c(b): #child
    def __init__(self,name):
        self.name = name
        print(self.name)
    def feature3(self):
        print("This is feature 3")
```

```python
cust3 = c("Innomatics")
cust3.feature1()
cust3.feature2()
cust3.feature3()
```

# calling constructors using super ()

```python
class a:
    def __init__(self, name):
        self.name = name
        print("A:", self.name)

    def feature1(self):
        print("this is feature1")


class b(a):
    def __init__(self, name):
        super().__init__(name)
        print("B:", self.name)

    def feature2(self):
        print("this is feature2")


class c(a):
    def __init__(self, name):
        super().__init__(name)
        print("C:", self.name)

    def feature3(self):
        print("This is feature 3")


class d(b, c):
    def __init__(self, name):
        super().__init__(name)
        print("D:", self.name)

    def feature4(self):
        print("This is feature 4")


class e(d):
    def __init__(self, name):
        super().__init__(name)
        print("E:", self.name)

    def feature5(self):
        print("This is feature 5")


obj = e("Ravi")
```
