```python
#find() function
name.find(" ")
```

```python
count_spaces=0
if name.find(" ") == -1:
    print("no space",0)
start_idx=0
while name[start_idx:].find(" ") != -1:
    count_spaces+=1
    start_idx+=name[start_idx:].find(" ")+1    
```
