# Recursion
- calling a function repetatively

```python
def factorial(x):
    if x<=1:
        return x
    else:
        return x* factorial(x-1)
factorial(4)
```

```python
def factorial(a):
    fact = 1
    for i in range(1,a+1):
        fact = fact * i
    return fact
print(factorial(5))
    
```
