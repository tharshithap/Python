# 1.Print the first half of a string.

```python

s="ABCDEFGHIJKLMNOP"
print(s[:8])
```

```python
# or
h = len(s)//2
print(s[:h])
```

# 2. Print the second half of a string.


```python
print(s[8:])
# or
h = len(s)//2
print(s[h:])
```

# 3. Extract the middle three characters.


```python
print(s[7:10])
#m or
h = len(s)//2
print(s[h-1] + s[h]+s[h+1])
```

# 4. Print every alternate character.


```python
print(s[::2])
```

#  5.Reverse a string using slicing

```python
print(s[::-1])
```
