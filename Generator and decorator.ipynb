--> In generator we should use yield 

```python
def func(num):
    square =[]
    for i in num:
        square.append(i**2)
    return square
```

```python
func([1,2,3,4,56])
```

```python
def gen_fun(num):
    for i in num:
        yield i**3
```

```python
gen=gen_fun([1,2,3,4,5,6])
```

```python
print(next(gen))
```

```python
print(next(gen))
```

```python
print(next(gen))
```
