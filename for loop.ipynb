# For loop
- used for iterating a loop multiple times
- 
**Syntax**

  for temporary_variable_name in variable:
  
        stmts(things to do)

```python
tasks = ["sleeping","freshesn up","scrolling phone","breakfast","classes","lunch","scrolling","Dinner","sleeping"]
for task in tasks:
    print(task)
```

```python
fruits=['apple','banana','grapes','orange','pineapple','watermelon']
for i in fruits:
    print(i)
```

# Using enumerate() in Python
- enumerate() helps both index and value during iteration

**Syntax:-** enumerate(iterable,start=0)
- It returns (index,item)

```python
tasks = ["sleeping","freshesn up","scrolling phone","breakfast","classes","lunch","scrolling","Dinner","sleeping"]
for index,value in enumerate(tasks,start=0):
    print(index,value)
```

# Nested for 

```python
for i in range(1,6,1):
    for j in range(1,6,1):
            print(j)
```

```python
for i in range(1,6,1):
    for j in range(1,6,1):
            print(i,j)
```

```python
for i in range(1,6,1):
    for j in range(1,6,1):
        for k in range(1,6,1):
            print(i,j,k)
```

```python
for i in range(9,10,1):
    for j in range(1,11,1):
        print(f'{i} x {j} == {i*j}')
```

```python
# Print 1 to 5 multiplication table
for i in range(1,6,1):
    for j in range(1,11,1):
        print(f'{i} x {j} == {i*j}')
    print("--------")
```

# Control Flow stmts
- **break** : It breaks the loop according to the condition
- **Pass** :  It allows the value
- **Continue** : It skip the value 

```python
# Break
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
for i in range(10):
    if i == 5:
        pass
    print(i)
```

```python
for i in range(10):
    if i == 5:
        continue
    print(i)
```
