```python
text = "Python is a general purpose language"
user_input = input("Enter a search word:")
start= text.index(user_input)
end = start + len(user_input)
print(f"The word{user_input} is starting from {start} to {end}")
```

```python
ext = "Python is a general purpose language"
user_input = input("Enter a search word:")
start= text.index(user_input)
end = start + len(user_input)
print(f"The word{user_input} is starting from {start} to {end}")
```

```python
ext = "Python is a general purpose language"
user_input = input("Enter a search word:")
start= text.find(user_input)
end = start + len(user_input)
print(f"The word{user_input} is starting from {start} to {end}")
```

```python
text = "Python is a general purpose language"
user_input = input("Enter a search word:")
if user_input in text:
    start= text.index(user_input)
    end = start + len(user_input)
    print(f"The word{user_input} is starting from {start} to {end}")
else:
    print(f"The word {user_input} is not found")
```
