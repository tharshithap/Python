# 3. Slicing
Given:

colors = ("Red", "Green", "Blue", "Yellow", "Black")

- Print the first three colors.
- Print the last two colors.
- Print every second color.




```python
c = ("Red", "Green", "Blue", "Yellow", "Black")
print("Frist three colors:",c[0:3])
print("Last two colors:",c[-2:])
print("Every second colors:",c[1::2])
```

# 13.What is the difference between remove() and discard()? 
- **remove()** : removes an element and raises a Error if the element is not found.
- **discard()**: removes an element but does not raise an error if the element is not found.

# 14.Explain the difference between union() and update().
- **Union()** : Returns a new set containing all unique elements from both sets.

- **Update()** : Adds all unique elements from another set to the existing set.
