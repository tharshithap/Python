## my append()

```python
def my_append(a,b):
    x = type(a)
    if type(a)!=list:
        print(f"{x} object has no attribute 'append' ")
    else:
        return a+b
```

```python
my_append([1,2,3,4],[6])
```

```python
my_append((1,2,3,4),(6))
```
