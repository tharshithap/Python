```python
### ..........................OPERATORS....................................####
## 1.Arithmetic operators

num1 = 12
num2 = 5


print("Addition:", num1 + num2)
print("Subtraction:", num1 - num2)
print("Multiplication:", num1 * num2)
print("Division:", num1 / num2)
print("Floor Division:", num1 // num2)
print("Remainder:", num1 % num2)
print("Power:", num1 ** num2)

##Square root

9**.5
```

```python
## 2.Logical operators

is_sunny = True
is_warm = False

print("Both sunny and warm:", is_sunny and is_warm)
print("Either sunny or warm:", is_sunny or is_warm)
print("Not sunny:", not is_sunny)
```

```python
#Assignment operators
x= 15
y = 6
print("Before assignment: x =", x, "y =", y)
x += y
print(" x =", x, "y =", y)
x -= y
print("x =", x, "y =", y)
x *= y
print("x =", x, "y =", y)
x /= y
print("x =", x, "y =", y)
x //=y
print("x =", x, "y =", y)
x %= y
print("x =", x, "y =", y)
```
