# List Comprehension

```python
l=[]
for i in range(1,10,1):
    l.append(i**2)
print(l)
```

```python
l=[i**2 for i in range(1,10,1)]
print(l)
```

```python
l=[i**.5 for i in range(1,10,1)]
print(l)
```

# if conditions in list comprehension
- **Syntax** [Expression for loop iterable if condition]

```python
l=[i for i in range(1,10,1) if i%2==0]
l
```

- **Syntax** [expression if condition else expression loop iterables]

```python
l = [ "Even" if  i%2==0 else "Odd" for i in range(1,30,1) ]
print(l)
```

```python
n = input()
l=[  i for i in n if i.lower() in "aeiou" ]
l
```

```python
n = input()
l=[ "Vowels" if i.lower() in "aeiou" else "conso" for i in n  ]
print(l)
```

# Nested for loop
- **Syntax**[[inner loop]outer loop]


```python
l = [[(i,j) for  j in range(1,10)] for i in range(1,3,1)]
l
```

```python
l = [[(i*j) for  j in range(1,10)] for i in range(1,3,1)]
l
```
