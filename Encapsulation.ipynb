# Encapsulation – Python Notes
**1. What is Encapsulation?**

Encapsulation means wrapping data and methods together inside a class and controlling access to the data.

In simple words:

Encapsulation = Data + Methods + Data Protection

```python
class Bank:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self,amount):
        self.balance += amount

    def check_balance(self):
        print(self.balance)
b = Bank(50000)
b.deposit(500)
b.check_balance()
```

# 2. Why Use Encapsulation?
Security: Protects internal object state.

Data Integrity: Enforces validation or business rules.

Simplicity: Provides a clear interface to interact with the object.

Maintainability: Changes to internal implementation don't affect external usage.

# 3. Access Modifiers in Python



Python mainly uses three levels:

Type	                          Syntax	                            Meaning
                  
Public	                            name	                            Can be accessed anywhere

Protected	                        name	                      Intended for class and child classes

Private	                            name	              Intended to be accessed only inside the class

# PUBLIC

```python
class Student:
    def __init__(self):
        self.name = "Tharshi"

s = Student()
print(s.name)
```

# PROTECTED

```python
class Student:
    def __init__(self):
        self._name = "Tharshi"

s = Student()
print(s._name)
```

# PRIVATE 

```python
class Student:
    def __init__(self):
        self.__name = "Tharshi"

    def display(self):
        print(self.__name)


s = Student()
s.display()
```

# 4. Encapsulation with Getter and Setter

 - A getter is used to read data.

 - A setter is used to change data.

```python
class Bank:
    def __init__(self, balance):
        
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def set_balance(self, balance):
        self.__balance = balance


b = Bank(5000)

print(b.get_balance())

b.set_balance(10000)

print(b.get_balance())
```

# 5. Real-Life Example

- Think about an ATM.

- You don't directly change your bank balance.

- Instead, you use:

    - Deposit
    - Withdraw
    - Check Balance

```python
class ATM:
    def __init__(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print(self.__balance)

    def withdraw(self, amount):
        if amount <= self.__balance:
            self.__balance -= amount
            print(self.__balance)
        else:
            print("Insufficient Balance")

    def check_balance(self):
        print("Balance:", self.__balance)
```

```python
Atm = ATM (50000)
Atm.deposit(500)
Atm.withdraw(300)
Atm.check_balance()
```
