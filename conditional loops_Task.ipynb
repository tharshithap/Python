# 1.What is a conditional statement in Python?
- A conditional statement in Python is a statement that checks a condition and executes a block of code only if the condition is True. It helps the program make decisions.

# 2. What is the difference between if, if-else, and if-elif-else?
**if** - Executes a if block of code if the condition is True, if false nothing returns

**else** - Executes a if block of code if the condition is True , otherwise the else block is executed

**if-elif-else** - Checks multiple conditions one by one. The first True condition is executed. If none are True, the else block runs.

# 16. Write a program to calculate grade based on marks:
- 90–100 → A
- 75–89 → B
- 60–74 → C
- 40–59 → D
- Below 40 → Fail


```python
marks = int(input("Enter your marks: "))
if marks >= 90 and marks <= 100:
    print("Grade: A")
elif marks >= 75 and marks <= 89:
    print("Grade: B")
elif marks >= 60 and marks <= 74:
    print("Grade: C")
elif marks >= 40 and marks <= 59:
    print("Grade: D")
elif marks >= 0 and marks < 40:
    print("Fail")
else:
    print("Invalid marks! Please enter marks between 0 and 100.")
```

#  17.Write a program to calculate a tax amount based on salary slabs. 
**Example Slabs**
- Salary ≤ ₹2,50,000 → No tax
- ₹2,50,001–₹5,00,000 → 5%
- ₹5,00,001–₹10,00,000 → 20%
- Above ₹10,00,000 → 30%


```python
salary = float(input("Enter your salary: ₹"))
if salary <= 250000:
    tax = 0
elif salary <= 500000:
    tax = salary * 0.05
elif salary <= 1000000:
    tax = salary * 0.20
else:
    tax = salary * 0.30
print("Salary: ₹", salary)
print("Tax Amount: ₹", tax)
```

# 18.Write a program to create a simple calculator using +, -, *, /. 


```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operator = input("Enter operator (+, -, *, /): ")
if operator == "+":
    print("Result =", num1 + num2)
elif operator == "-":
    print("Result =", num1 - num2)
elif operator == "*":
    print("Result =", num1 * num2)
elif operator == "/":
        print("Result =", num1 / num2)
else:
    print("Invalid operator!")
```

# 19.Write a program to check whether a given character is:
- Uppercase
- Lowercase
- Digit
- Special character


```python
ch = input("Enter a character: ")
if ch.isupper():
    print("Uppercase Letter")
elif ch.islower():
    print("Lowercase Letter")
elif ch.isdigit():
    print("Digit")
else:
    print("Special Character")
```
