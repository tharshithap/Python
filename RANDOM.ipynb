# Syntax: np.nditer(array_name)

```python
for i in np.nditer(a1):
    print(i)
```

```python
# create 2d array/matrix and check the values is even or odd
a1 = np.arange(1,5,1).reshape(2,2)
for i in np.nditer(a1):
    if i%2==0:
        print("even",i)
    else:
        print("odd",i)
      
```

```python
# create 3d array/matrix and check the values is even or odd
a1 = np.arange(1,10,1).reshape(3,3)
for i in np.nditer(a1):
    if i%2==0:
        print("even",i)
    else:
        print("odd",i)
      
```

```python
a1 = np.array([[1,2],[3,4]])
for i in np.nditer(a1,order='f'):
    print(i)
```
