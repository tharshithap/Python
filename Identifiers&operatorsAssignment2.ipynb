1. Valid and invalid identifiers
   
Valid identifiers
my_variable = 25

counter = 40.7

_total = "Hello"

Name123 = 136

myFunction = True

_data = False

Invalid identifiers
123abc = 34

my variable = 6.7

@count = "hi"

for = "True"

if_condition? = "Innomatics"

class = "Hyderabad"










2.Why they are invalid

-123abc starts with a digit, and identifiers cannot begin with a number.

-my variable has a space, and identifiers cannot contain spaces.

-@count has @, and special characters other than underscore are not allowed.

-for is a Python keyword, so it cannot be used as an identifier.

-if_condition? contains ?, which is not allowed in identifiers.

-class is also a Python keyword, so it is invalid as an identifier.

Arrays and lists
Differences:
-A list is a built-in Python data structure that can store mixed data types.

-An array is usually used for storing same-type values and is more memory-efficient for numeric work.

-Lists are more flexible, while arrays are better for fast numerical processing.

When to use lists
-Use lists when you need different data types together.

-Use lists when you want easy insertion, deletion, or general-purpose storage.

Example: ["Tharshitha", 21, True]

When to use arrays
-Use arrays when you are handling large numeric data.

-Use arrays when you need faster computation.

Example: numeric calculations, machine learning, scientific computing.

Justification:

-For normal Python programming, lists are usually better because they are flexible.

-For heavy numeric operations, arrays are better because they are more efficient.
