# POLYMORPHISM 
- Polymorphism means "one name, many forms."

- In Python, the same method name can behave differently depending on the object.

```python
a = 3
b= 4
print(a+b)
```

```python
a = "3"
b = "4"
print(a+b)
```

```python
l= [1,2,3,4]
t = [23,4,5,6,7]
print(len(l))
```

```python
print(len(t))
```

```python
class car:
    def __init__(self):
        pass
    def horn(self):
        print("car is honking")
class bike:
    def __init__(self):
        pass
    def horn(self):
        print("bike is honking")
class bus:
    def __init__(self):
        pass
    def horn(self):
        print("bus is honking")
for i in (car(),bike(),bus()):
    i.horn()

```

```python
class vechile:
    def __init__(self):
        pass
    def horn(self):
        print("vechile is honking")
class car(vechile):
    def __init__(self):
        pass
    def horn(self):
        print("car is honking")
class bus(vechile):
    def __init__(self):
        pass

for i in (car(),bus()):
    i.horn()
           
```
